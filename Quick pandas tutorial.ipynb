## Working with data, sorting, filtering, saving.

Sorting the data by gender_id and title in ascending order

```python
# Sorting the data set by gender_id and title in ascending order in a new dataframe
genre_sort = google_data.sort_values(by=['genre_id', "title"], ascending=True)
# Reset the index for the new sorted data
genre_sort.reset_index(drop=True, inplace=True)
# Show the head of the sorted data set
genre_sort.head()
```

Note that the index now starts at 0 and the values are ordered by genre_id and then by title name.

Filter data by creating a new dataframe by selecting and sorting the data with more than 10 million 5-stars rating.
It will need to create a new dataframe using a boolean function. The boolean command will be used to select the data with more than 10 million 5-stars from the dataset.

```python
# Select the rows with more than 10E+7 5-stars
top_5_stars = google_data[google_data['rating_five_star'] > 1E+7]
```

```python
# Sort the selected data and re-index
top_5_stars = top_5_stars.sort_values(by=['rating_five_star'], ascending=False)
top_5_stars.reset_index(drop=True, inplace=True)
```

```python
# Verify data selection and index reset
top_5_stars[['title','rating_five_star']]
```

## Correlations between columns

```python
# Selection of columns with numbers to find correlations among them.
top_5_stars[['ratings', 'score', 'min_installs']].corr()
```
